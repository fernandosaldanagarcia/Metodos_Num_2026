### Ejemplo relacionado con métodos numéricos

El error absoluto de una aproximación $\widetilde{x}$ a un valor exacto $x$ se define como

$$
E_{\mathrm{abs}} = |x-\widetilde{x}|.
$$

El error relativo se define como

$$
E_{\mathrm{rel}}
=
\frac{|x-\widetilde{x}|}{|x|},
\qquad x\neq 0.
$$


````python
import math

valor_exacto = math.pi
aproximacion = 22 / 7

error_absoluto = abs(valor_exacto - aproximacion)
error_relativo = error_absoluto / abs(valor_exacto)

print("Valor exacto:", valor_exacto)
print("Aproximación:", aproximacion)
print("Error absoluto:", error_absoluto)
print("Error relativo:", error_relativo)
````

Las funciones permiten organizar y reutilizar cálculos.


````python
def f(x):
    """Calcula f(x) = x^3 - x - 2."""
    return x**3 - x - 2

print(f(1))
print(f(2))
````


Una función puede recibir varios argumentos.


````python
def error_relativo(valor_exacto, aproximacion):
    """Calcula el error relativo de una aproximación."""
    if valor_exacto == 0:
        raise ValueError("El valor exacto debe ser distinto de cero.")
    return abs(valor_exacto - aproximacion) / abs(valor_exacto)

error_relativo(np.pi, 22/7)
````

## 11. Impresión con formato

Las *f-strings* permiten insertar valores dentro de texto.


````python
x_aprox = 1.41421356
iteraciones = 8

print(f"Aproximación: {x_aprox}")
print(f"Aproximación con 4 decimales: {x_aprox:.4f}")
print(f"Número de iteraciones: {iteraciones}")
````


## 14. Ejemplo: método de bisección

Consideremos la ecuación

$$
f(x)=x^3-x-2=0.
$$

El método de bisección parte de un intervalo $[a,b]$ tal que

$$
f(a)f(b)<0.
$$

En cada iteración se calcula

$$
c=\frac{a+b}{2}
$$

y se conserva el subintervalo donde ocurre el cambio de signo.


````python
def biseccion(f, a, b, tolerancia=1e-8, max_iter=100):
    """Aproxima una raíz de f en [a,b] mediante bisección."""
    fa = f(a)
    fb = f(b)

    if fa * fb >= 0:
        raise ValueError("f(a) y f(b) deben tener signos opuestos.")

    historial = []

    for k in range(max_iter):
        c = (a + b) / 2
        fc = f(c)
        error_estimado = (b - a) / 2

        historial.append((k, a, b, c, fc, error_estimado))

        if abs(fc) < tolerancia or error_estimado < tolerancia:
            return c, historial

        if fa * fc < 0:
            b = c
            fb = fc
        else:
            a = c
            fa = fc

    return c, historial


raiz, historial = biseccion(f, 1, 2)

print(f"Raíz aproximada: {raiz:.10f}")
print(f"f(raíz): {f(raiz):.4e}")
print(f"Iteraciones: {len(historial)}")
````


Podemos mostrar las primeras iteraciones en forma de tabla.


````python
import pandas as pd

tabla = pd.DataFrame(
    historial,
    columns=["k", "a", "b", "c", "f(c)", "error estimado"]
)

tabla.head(10)
````


También podemos visualizar la disminución del error estimado.


````python
iteraciones = tabla["k"]
errores = tabla["error estimado"]

plt.figure(figsize=(7, 4))
plt.semilogy(iteraciones, errores, marker="o")
plt.xlabel("Iteración")
plt.ylabel("Error estimado")
plt.title("Convergencia del método de bisección")
plt.grid(True)
plt.show()
````
