$A=LU$

sin permutaciones de filas, con $L$ triangular inferior y diagonal unitaria, y $U$ triangular superior, si y solo si todos los menores principales principales de $A$ son distintos de cero.

$A_k=A(1:k,1:k)$,

$\det(A_k)\neq 0,\qquad k=1,\ldots,n$.
