## 1. Importar NumPy

Por convención, NumPy se importa utilizando el alias `np`.


````python
import numpy as np
````


## 2. Definición de vectores

En NumPy, un vector puede representarse mediante un arreglo unidimensional.

Por ejemplo, el vector

$$
x =
\begin{pmatrix}
1\\
2\\
-1
\end{pmatrix}
$$

se define de la siguiente manera:


````python

x = np.array([1, 2, -1])

print("Vector x:")
print(x)
print("Dimensión del arreglo:", x.shape)
````


### Vector fila y vector columna

Un arreglo de dimensión `(3,)` no distingue explícitamente entre vector fila y vector columna.

Podemos crear un vector fila con forma `(1,3)`:


````python

x_fila = np.array([[1, 2, -1]])

print(x_fila)
print("Dimensión:", x_fila.shape)
````

Y un vector columna con forma `(3,1)`:

````python

x_columna = np.array([[1], [2], [-1]])

print(x_columna)
print("Dimensión:", x_columna.shape)
````


## 3. Definición de matrices

La matriz

$$
A =
\begin{pmatrix}
2 & -1 & 0\\
1 & 3 & 2\\
0 & 1 & 4
\end{pmatrix}
$$

se define como un arreglo bidimensional.


````python

A = np.array([
    [2, -1, 0],
    [1,  3, 2],
    [0,  1, 4]
])

print("Matriz A:")
print(A)
print("Dimensión:", A.shape)
````


Algunas matrices especiales pueden construirse directamente con NumPy.


````python

I = np.eye(3)          # matriz identidad
Z = np.zeros((3, 3))   # matriz de ceros
U = np.ones((2, 4))    # matriz de unos

print("Identidad:")
print(I)

print("\nMatriz de ceros:")
print(Z)

print("\nMatriz de unos:")
print(U)
````


## 4. Operaciones con vectores

Sean

$$
x=(1,2,-1), \qquad y=(3,0,4).
$$

La suma y la resta se realizan componente a componente.


````python

x = np.array([1, 2, -1])
y = np.array([3, 0, 4])

print("x + y =", x + y)
print("x - y =", x - y)
````

La multiplicación de un vector por un escalar también se realiza componente a componente.

````python

alpha = 3

print("3x =", alpha * x)
````


### Producto componente a componente

El operador `*` entre dos vectores de NumPy realiza una multiplicación componente a componente:

$$
x*y=(x_1y_1,\ldots,x_ny_n).
$$


````python
print("Producto componente a componente:", x * y)
````


### Producto punto o producto escalar

El producto punto de dos vectores de $\mathbb{R}^n$ es

$$
x^\top y = \sum_{i=1}^n x_i y_i.
$$

En NumPy se utiliza `np.dot(x,y)` o el operador `@`.


````python

producto_punto_1 = np.dot(x, y)
producto_punto_2 = x @ y

print("np.dot(x, y) =", producto_punto_1)
print("x @ y =", producto_punto_2)
````


## 5. Operaciones con matrices

Definamos dos matrices:

$$
A=
\begin{pmatrix}
1&2\\
3&4
\end{pmatrix},
\qquad
B=
\begin{pmatrix}
2&0\\
-1&5
\end{pmatrix}.
$$


````python

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [2, 0],
    [-1, 5]
])

print("A =")
print(A)

print("\nB =")
print(B)
````

La suma y resta se realizan elemento a elemento.

````python

print("A + B =")
print(A + B)

print("\nA - B =")
print(A - B)
````


### Multiplicación elemento a elemento

El operador `*` no calcula el producto matricial. Multiplica entradas correspondientes.


````python

print("A * B =")
print(A * B)
````


### Producto matricial

El producto matricial se calcula con el operador `@` o mediante `np.matmul`.


````python

print("A @ B =")
print(A @ B)

print("\nB @ A =")
print(B @ A)
````


Observe que, en general,

$$
AB \neq BA.
$$



## 6. Producto matriz-vector

Si $A\in\mathbb{R}^{m\times n}$ y $x\in\mathbb{R}^n$, entonces

$$
Ax\in\mathbb{R}^m.
$$

El producto matriz-vector se realiza con el operador @
Nota que al realizar la multiplicación, NumPy automáticamente
detecta la dimensión correcto del vector (por ende no es necesario
escribirlo como vector columna).


````python

A = np.array([
    [2, -1],
    [0,  3],
    [1,  4]
])

x = np.array([2, 1])

Ax = A @ x

print("A =")
print(A)
print("\nx =", x)
print("\nAx =", Ax)
````


## 7. Transpuesta de una matriz

La transpuesta de $A$, denotada por $A^\top$, se obtiene usando A.T en Python.


````python

A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("A =")
print(A)

print("\nTranspuesta de A:")
print(A.T)
````


## 8. Determinante

Para una matriz cuadrada \(A\), el determinante se calcula con

```python
np.linalg.det(A)
```

Considere

$$
A=
\begin{pmatrix}
2&1\\
3&4
\end{pmatrix}.
$$

Teóricamente,

$$
\det(A)=2(4)-1(3)=5.
$$


````python

A = np.array([
    [2, 1],
    [3, 4]
])

det_A = np.linalg.det(A)

print("det(A) =", det_A)
````


Debido a que los cálculos se realizan en aritmética de punto flotante, el resultado podría aparecer como un número muy cercano al valor exacto.

Para comprobar si una matriz es invertible, se puede verificar si su determinante es distinto de cero. Sin embargo, en cálculo numérico normalmente es preferible utilizar otras herramientas, como el rango, la factorización o el número de condición.



## 9. Eigenvalores e eigenvectores

Un número $\lambda$ es un eigenvalor de una matriz cuadrada $A$ si existe un vector no nulo $v$ tal que

$$
Av=\lambda v.
$$

El vector $v$ se denomina eigenvector asociado a $\lambda$.

En NumPy se utiliza:

```python
np.linalg.eig(A)
```

para calcular eigenvectores y también eigenvalores. 

En caso de que se quieran obtener solo los eigenvalores se utiliza

```python
np.linalg.eigvals(A)
```

````python

A = np.array([
    [2, 1],
    [1, 2]
])

eigenvalores, eigenvectores = np.linalg.eig(A)

print("Eigenvalores:")
print(eigenvalores)

print("\nEigenvectores:")
print(eigenvectores)
````


Cada columna de la matriz `eigenvectores` es un eigenvector. La columna $j$ corresponde al eigenvalor `eigenvalores[j]`.

Podemos verificar numéricamente la relación

$$
Av_j=\lambda_jv_j.
$$


````python

j = 0

lambda_j = eigenvalores[j]
print("Primer eigenvalor", lambda_j)

v_j = eigenvectores[:, j]
print("\nPrimer eigenvector", v_j)

print("\nA v_j =")
print(A @ v_j)

print("\nlambda_j v_j =")
print(lambda_j * v_j)

````


### Matrices simétricas

Para matrices reales simétricas es recomendable utilizar `np.linalg.eigh`, ya que está diseñada específicamente para matrices simétricas o hermíticas (en este caso los eigenvalores son siempre reales).


````python

A_sim = np.array([
    [2, 1],
    [1, 3]
])

valores, vectores = np.linalg.eigh(A_sim)

print("Valores propios:")
print(valores)

print("\nVectores propios:")
print(vectores)
````


## 10. Radio espectral

El radio espectral de una matriz $A$ es

$$
\rho(A)=\max_{\lambda\in sp(A)}|\lambda|,
$$

donde $sp(A)$ es el espectro de A (conjunto de valores propios de $A$).


````python

A = np.array([
    [0, 1],
    [-2, 3]
])

eigenvalores = np.linalg.eigvals(A)
radio_espectral = np.max(np.abs(eigenvalores))

print("Valores propios:", eigenvalores)
print("Radio espectral:", radio_espectral)
````
