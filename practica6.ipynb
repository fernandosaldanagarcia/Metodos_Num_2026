## Problema 1. Método de Jacobi

Considera el sistema lineal
$$
A\mathbf{x}=\mathbf{b}, \qquad
A=\begin{pmatrix}
4&-1&0\\
-1&4&-1\\
0&-1&3
\end{pmatrix},
\qquad
\mathbf{b}=\begin{pmatrix}15\\10\\10\end{pmatrix}.
$$
### Fórmulas de iteración

Para un sistema $A\mathbf{x}=\mathbf{b}$, se despeja la componente $x_i$ de cada ecuación. El método de Jacobi calcula todas las componentes nuevas usando únicamente los valores de la iteración anterior:
$$
x_i^{(k+1)}
=\frac{1}{a_{ii}}\left(b_i-\sum_{j\ne i}a_{ij}x_j^{(k)}\right),
\qquad i=1,\ldots,n.
$$

En forma matricial, si $A=D-E-F$, donde $D$ es la diagonal y $E,F$ contienen los valores negativos de las partes estrictamente inferior y superior, respectivamente de A,

$$
\mathbf{x}^{(k+1)}
=D^{-1}\left((E+F)\mathbf{x}^{(k)} + \mathbf{b}\right).
$$

## Problema 2. Método de Gauss-Seidel

Considera el sistema

$$
A\mathbf{x}=\mathbf{b},\qquad
A=\begin{pmatrix}
4&-1&0\\
-1&4&-1\\
0&-1&3
\end{pmatrix},
\qquad
\mathbf{b}=\begin{pmatrix}15\\10\\10\end{pmatrix}.
$$

### Fórmulas de iteración

Gauss-Seidel utiliza inmediatamente los valores nuevos que ya se han calculado durante la iteración actual:

$$
x_i^{(k+1)}
=\frac{1}{a_{ii}}\left(
b_i-\sum_{j<i}a_{ij}x_j^{(k+1)}
-\sum_{j>i}a_{ij}x_j^{(k)}
\right).
$$

En forma matricial, usando la descomposición $A=D-E-F$,

$$
\mathbf{x}^{(k+1)}
=(D-E)^{-1}\left(F\mathbf{x}^{(k)} + \mathbf{b}\right).
$$

## Problema 3. Método de Relajación

$$
x_i^{(k+1)}
=
(1-\omega)x_i^{(k)}
+
\frac{\omega}{a_{ii}}
\left(
b_i
-\sum_{j=1}^{i-1}a_{ij}x_j^{(k+1)}
-\sum_{j=i+1}^{n}a_{ij}x_j^{(k)}
\right),
\quad i=1,\ldots,n.
$$

Que en forma matricial equivale a:

$$
x^{(k+1)}
=
\left(\dfrac{D}{\omega} - E\right)^{-1}
\left[
\left(\dfrac{1-\omega}{\omega} D + F\right)x^{(k)}
+ b
\right].
$$
