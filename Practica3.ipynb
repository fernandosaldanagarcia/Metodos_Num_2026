
## Normas de vectores

Sea

$$
x=(x_1,\ldots,x_n).
$$

### Norma 1

$$
\|x\|_1=\sum_{i=1}^n |x_i|.
$$

### Norma 2 o norma euclidiana

$$
\|x\|_2=
\left(\sum_{i=1}^n |x_i|^2\right)^{1/2}.
$$

### Norma infinito

$$
\|x\|_\infty=\max_i |x_i|.
$$

en Python dichas normas se pueden calcular como se indica en el siguiente bloque de código.

````python

x = np.array([1, 0, 3, -2])

norma_1 = np.linalg.norm(x, ord=1)
norma_2 = np.linalg.norm(x, ord=2)
norma_inf = np.linalg.norm(x, ord=np.inf)

print("x =", x)
print("Norma 1:", norma_1)
print("Norma 2:", norma_2)
print("Norma infinito:", norma_inf)
````


Podemos calcular las normas de forma manual de la siguiente manera:


````python
x = np.array([1, 0, 3, -2])

norma_1_manual = np.sum(np.abs(x))
norma_2_manual = np.sqrt(np.sum(x**2))
norma_inf_manual = np.max(np.abs(x))

print("Norma 1 calculada manualmente:", norma_1_manual)
print("Norma 2 calculada manualmente:", norma_2_manual)
print("Norma infinito calculada manualmente:", norma_inf_manual)
````


## Normas de matrices

Para una matriz $A=(a_{ij})\in\mathbb{C}^{m\times n}$, algunas normas importantes son las siguientes.

### Norma matricial 1

$$
\|A\|_1 =
\max_{1\leq j\leq n}
\sum_{i=1}^m |a_{ij}|.
$$

Es decir, es la máxima suma absoluta de las columnas.

### Norma matricial infinito

$$
\|A\|_\infty =
\max_{1\leq i\leq m}
\sum_{j=1}^n |a_{ij}|.
$$

Es decir, es la máxima suma absoluta de las filas.

### Norma matricial 2

$$
\|A\|_2 =
\sqrt{\rho(A^T A)}.
$$

Es decir, la raíz cuadrada del radio espectral del producto $A^T A$ (si $A$ es compleja cambiar $A^T$ por $A^*$).

Los mismos comandos que cálculan normas vectoriales pueden utilizarce para cálcular normas matriciales.


````python

A = np.array([
    [1, -2,  0],
    [3,  1, -1]
])

norma_A_1 = np.linalg.norm(A, ord=1)
norma_A_2 = np.linalg.norm(A, ord=2)
norma_A_inf = np.linalg.norm(A, ord=np.inf)

print("A =")
print(A)

print("\nNorma matricial 1:", norma_A_1)
print("Norma matricial 2:", norma_A_2)
print("Norma matricial infinito:", norma_A_inf)
````


Podemos cálcular las normas 1 e infinito "manualmente" calculando las sumas absolutas de columnas y filas.


````python

sumas_columnas = np.sum(np.abs(A), axis=0) # suma cada columna y guarda el resultado en un vector
sumas_filas = np.sum(np.abs(A), axis=1) # suma cada fila y guarda el resultado en un vector

print("Sumas absolutas por columna:", sumas_columnas)
print("Máxima suma por columna:", np.max(sumas_columnas))

print("\nSumas absolutas por fila:", sumas_filas)
print("Máxima suma por fila:", np.max(sumas_filas))
````


##  Norma de Frobenius

La norma de Frobenius de una matriz es

$$
\|A\|_F =
\left(
\sum_{i=1}^m\sum_{j=1}^n |a_{ij}|^2
\right)^{1/2}.
$$

Puede interpretarse como la norma euclidiana del vector formado por todas las entradas de la matriz.

Se cálcula de la siguiente manera.


````python

A = np.array([
    [1, -2, 0],
    [3,  1, 4]
])

norma_F = np.linalg.norm(A, ord="fro")

print("A =")
print(A)
print("\nNorma de Frobenius:", norma_F)
````

La norma de Frobenius se puede calcular manualmente de la siguiente forma:

````python

norma_F_manual = np.sqrt(np.sum(A**2))

print("Norma de Frobenius calculada manualmente:", norma_F_manual)
````
